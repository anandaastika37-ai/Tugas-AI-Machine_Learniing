## **Data Understanding**

```python
import pandas as pd

data = pd.read_csv('../data/heart.csv')
dataFrame = pd.DataFrame(data)


class UnderstandingData:
    def __init__(self , dataFrame):
        self.df = dataFrame
    def checkStructure(self):
        print("DATASET INFORMATION")
        self.df.info()
        print(f"\nJumlah baris dan kolom : {self.df.shape}")
        print(f"Jumlah seluruh data   : {self.df.size}")
        print(f"\nDaftar kolom   : {self.df.columns.tolist()}")
        print(f"Jumlah dimensi : {self.df.ndim}")
    def checkData(self):
        display(self.df.head(10)) #10 data teratas
        display(self.df.tail(10)) #10 data paling bawah
        display(self.df.sample(10, random_state=42)) #10 data acak
        display(self.df[["age" , 'thalach']]) # kolom age & thalach
        display(self.df.loc[self.df["age"] > 50]) # filter umur lebih dari 50
        largest = self.df.nlargest(5, "chol").reset_index(drop=True)
        smallest = self.df.nsmallest(5, "chol").reset_index(drop=True)
        display(largest / smallest)
    def dataType(self):
        print("DATA TYPES ANALYSIS")
        print("\n1. Tipe Data Setiap Kolom")
        print(self.df.dtypes)
        print("\n2. Distribusi Tipe Data")
        print(self.df.dtypes.value_counts())
        print("\n3. Kolom Numerikal")
        print(self.df.select_dtypes(include=["number"]).columns.tolist())

    def missingValue(self):
        missing_count = self.df.isnull().sum()
        totalMissing , ukuran = missing_count.sum() , self.df.size
        print(f"{totalMissing / ukuran * 100:.2f}%")
        print(f"{len(self.df.dropna())} dari  {len(self.df)}")
        print(self.df[self.df.isnull().any(axis=1)]) 
        print(pd.DataFrame({"jumlah": missing_count,"persen": (missing_count / len(self.df) * 100).round(2)}).sort_values("persen", ascending=False))
    def duplicatedValue(self):
        print(self.df.duplicated().sum())
        duplicate_count = self.df.duplicated().sum()
        duplicate_mask = self.df.duplicated(keep=False)
        duplicate_rows = self.df[duplicate_mask]
        unique_count = self.df.drop_duplicates()
        print(unique_count.index.tolist())
        print(duplicate_rows.index.tolist())
        duplicates = self.df[self.df.duplicated(keep=False)]
        display(duplicates.sort_values(by=self.df.columns.tolist()))
        print(self.df.duplicated().sum() / len(self.df) * 100)
        display(duplicate_rows.sort_values(by=self.df.columns.tolist()))
        print(unique_count["target"].value_counts())
        print(self.df['target'].value_counts())
    def StatistikDeskriptif(self):
        print(self.df.describe())




init = UnderstandingData(dataFrame)
# print(init.checkData())
# print(init.checkStructure())
# print(init.dataType())
# print(init.missingValue())
# print(init.duplicatedValue())
print(init.StatistikDeskriptif())
```
